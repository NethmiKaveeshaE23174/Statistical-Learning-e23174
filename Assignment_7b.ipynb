# Comprehensive Guide to Gaussian Mixture Models: From Analytical Foundations to Financial Segmentation

This notebook presents a full theoretical derivation of the Gaussian Mixture Model (GMM) optimized via the Expectation-Maximization (EM) algorithm, followed by an interactive production-grade implementation using `scikit-learn` and `Plotly` on financial credit card behavior data.

## Part 1: Deriving the Marginal Density

We wish to determine the marginal probability density $p(x_i)$ of an observed data point $x_i \in \mathbb{R}^d$. The latent cluster allocation variable $C_i$ can take any integer value in $\{1, \dots, K\}$. By applying the **law of total probability** across all mutually exclusive cluster assignments, we integrate out the discrete hidden state:

$$p(x_i) = \sum_{k=1}^K p(x_i, C_i = k)$$

Applying the definition of conditional probability ($p(A, B) = p(A \mid B)P(B)$), we can write:

$$p(x_i) = \sum_{k=1}^K p(x_i \mid C_i = k) P(C_i = k)$$

Substituting the prior cluster membership probabilities $P(C_i = k) = \phi_k$ and the conditional multivariate Gaussian densities $X_i \mid C_i = k \sim \mathscr{N}(\mu_k, \Sigma_k)$, we arrive precisely at:

$$p(x_i) = \sum_{k=1}^K \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k)$$

### Explanation
This formulation is termed a **Gaussian mixture density** because it models a complex, multimodal global density function by blending ("mixing") $K$ separate, distinct linear component densities. The parameters $\phi_k$ represent the **mixing proportions (or weights)**. They act as a discrete probability distribution over the sub-populations, satisfying $\phi_k \ge 0$ and $\sum_{k=1}^K \phi_k = 1$.

---

## Part 2: Deriving the Posterior Cluster Probability

Given a fixed, observed data point $X_i = x_i$, we use **Bayes' rule** to update our prior belief regarding which cluster generated this observation. Bayes' rule states:

$$P(C_i = k \mid X_i = x_i) = \frac{p(X_i = x_i \mid C_i = k)P(C_i = k)}{p(x_i)}$$

Expanding the denominator using the marginal density derived via the law of total probability in Part 1, we obtain the general structural equation:

$$P(C_i = k \mid X_i = x_i) = \frac{p(X_i = x_i \mid C_i = k)P(C_i = k)}{\sum_{j=1}^K p(X_i = x_i \mid C_i = j)P(C_i = j)}$$

Substituting our functional forms ($P(C_i = k) = \phi_k$ and the Gaussian evaluation $\mathscr{N}(x_i \mid \mu_k, \Sigma_k)$) yields the explicit posterior expression:

$$\gamma_{ik} = P(C_i = k \mid X_i = x_i) = \frac{\phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k)}{\sum_{j=1}^K \phi_j \mathscr{N}(x_i \mid \mu_j, \Sigma_j)}$$

### Interpretation
The quantity $\gamma_{ik}$ represents the **responsibility** that cluster $k$ assumes for generating data point $x_i$. It is interpreted as a posterior probability because it updates the unconditional prior probability $\phi_k$ after absorbing the empirical evidence provided by the coordinates of $x_i$. Notice that for any given point $i$, $\gamma_{ik} \ge 0$ and $\sum_{k=1}^K \gamma_{ik} = 1$, forming a valid posterior probability distribution across the latent clusters.

---

## Part 3: One-Hot Encoding of the Latent Cluster Variable

Let $Z_i = [Z_{i1}, Z_{i2}, \dots, Z_{iK}]^T$ be a one-hot encoded indicator vector where $Z_{ik} = 1$ if $C_i = k$ and $0$ otherwise. Because $Z_{ik}$ is a binary random variable, its conditional expectation given $X_i = x_i$ is evaluated directly from its discrete probability distribution:

$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = \left(1 \cdot P(Z_{ik} = 1 \mid X_i = x_i)\right) + \left(0 \cdot P(Z_{ik} = 0 \mid X_i = x_i)\right)$$

Since the event $Z_{ik} = 1$ is perfectly equivalent to the event $C_i = k$, it follows that:

$$\mathbb{E}[Z_{ik} \mid X_i = x_i] = P(C_i = k \mid X_i = x_i) = \gamma_{ik}$$

Extending this element-wise expectation to the full structural vector $Z_i$ yields:

$$\mathbb{E}[Z_i \mid X_i = x_i] = \begin{bmatrix} \mathbb{E}[Z_{i1} \mid X_i = x_i] \\ \mathbb{E}[Z_{i2} \mid X_i = x_i] \\ \vdots \\ \mathbb{E}[Z_{iK} \mid X_i = x_i] \end{bmatrix} = \begin{bmatrix} \gamma_{i1} \\ \gamma_{i2} \\ \vdots \\ \gamma_{iK} \end{bmatrix}$$

### Conclusion
This proves that the **soft cluster assignment vector** in GMMs is mathematically equivalent to the **conditional expectation of the hidden state indicator vector** given the observed data features.

---

## Part 4: From Soft Assignment to Hard Clustering

### Soft Clustering
Soft clustering avoids making a definitive decision about cluster assignment. Instead, it maintains a continuous distribution vector $\mathbb{E}[Z_i \mid X_i = x_i] = [\gamma_{i1}, \dots, \gamma_{iK}]^T$. This captures the model's structural uncertainty, reflecting how a single data point can fractionally belong to multiple overlapping groups.

### Hard Clustering
Hard clustering forces a strict, deterministic assignment. It converts the continuous responsibility vector into a single discrete label by applying the Maximum A Posteriori (MAP) decision rule:

$$\widehat{C}_i = \arg\max_{1 \le k \le K} \gamma_{ik}$$

In this framework, soft clustering maps an observation to a position on a continuous simplex, while hard clustering maps it directly to the closest vertex of that simplex, discarding information about surrounding ambiguity.

---

## Part 5: Conditional Expectation of the Observation Given the Cluster

By definition of the conditional expectation for a continuous random variable:

$$\mathbb{E}[X_i \mid C_i = k] = \int_{\mathbb{R}^d} x_i \, p(x_i \mid C_i = k) \, dx_i$$

Since $X_i \mid C_i = k \sim \mathscr{N}(\mu_k, \Sigma_k)$, the density function under the integral is exactly that of a symmetric multivariate Gaussian centered at $\mu_k$. By integration properties of the first moment of a Gaussian distribution:

$$\mathbb{E}[X_i \mid C_i = k] = \mu_k$$

### Interpretation
The vector $\mu_k$ represents the true coordinate center (mean location) of component $k$ within the feature space $\mathbb{R}^d$.

### Comparison of Intentions
*   **$\mathbb{E}[Z_i \mid X_i = x_i]$**: This expectation operates over the **latent cluster space** ($1 \times K$). It treats the observation as fixed and computes the posterior probability profile over the hidden states. It answers: *"Given this specific location in space, what is the probability profile of the clusters that could have generated it?"*
*   **$\mathbb{E}[X_i \mid C_i = k]$**: This expectation operates over the **empirical feature space** ($1 \times d$). It treats the cluster identity as fixed and isolates its spatial center. It answers: *"If we assume a point belongs to cluster $k$, what do we expect its average feature values to be?"*

## Part 6: The Complete-Data Likelihood

If the hidden states $z_i$ were perfectly observed alongside $x_i$, the joint density of the complete data sequence would be written using the product over all points and components:

$$p(x_1, \dots, x_n, z_1, \dots, z_n) = \prod_{i=1}^n \prod_{k=1}^K \left[ \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}}$$

Taking the natural logarithm of both sides transforms the products into linear sums:

$$\ell_c = \log \left( \prod_{i=1}^n \prod_{k=1}^K \left[ \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}} \right)$$

$$\ell_c = \sum_{i=1}^n \sum_{k=1}^K \log \left( \left[ \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}} \right)$$

Bringing the exponent $z_{ik}$ out front as a linear multiplier yields the final form:

$$\ell_c = \sum_{i=1}^n \sum_{k=1}^K z_{ik} \left[ \log \phi_k + \log \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

### Optimization Advantage
If $z_{ik}$ were known constants, maximizing $\ell_c$ would be trivial. The double summation allows the optimization problem to **decouple completely** across the $K$ distinct components. Instead of dealing with a log-of-a-sum (as in the marginal log-likelihood), we maximize independent terms for each component. The parameters $(\mu_k, \Sigma_k)$ could be updated using standard closed-form maximum likelihood estimation (MLE) formulas, isolating only those data samples where $z_{ik} = 1$.

---

## Part 7: The EM Interpretation

Because $Z_i$ is unobserved, we cannot compute $\ell_c$ directly. The **Expectation-Maximization (EM) algorithm** solves this by calculating the conditional expectation of the complete-data log-likelihood with respect to the posterior distribution of the latent variables, given the observed data and current parameter values $\theta^{(t)}$.

In the **E-step**, we replace the unknown indicator variables $z_{ik}$ with their conditional expectations:

$$z_{ik} \quad\leadsto\quad \mathbb{E}[Z_{ik} \mid X_i = x_i, \theta^{(t)}] = \gamma_{ik}$$

Substituting these expected values into the complete log-likelihood yields the surrogate objective function $Q$:

$$Q(\theta \mid \theta^{(t)}) = \sum_{i=1}^n \sum_{k=1}^K \gamma_{ik} \left[ \log \phi_k + \log \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

### Interpretation
The E-step can be seen as a **conditional assignment update**. Instead of forcing a definitive choice, it uses our current parameters to distribute fractional ownership of each data point across all $K$ clusters based on their posterior probabilities.

---

## Part 8: Parameter Updates (M-Step)

The **M-step** updates our parameter estimates by maximizing the surrogate function $Q$ with respect to $\phi_k$, $\mu_k$, and $\Sigma_k$.

### 1. Mixing Weights ($\phi_k$)
We maximize $Q$ subject to the constraint $\sum_{k=1}^K \phi_k = 1$ using a Lagrange multiplier $\lambda$:

$$\mathcal{L}(\phi, \lambda) = \sum_{i=1}^n \sum_{k=1}^K \gamma_{ik} \log \phi_k + \lambda \left( 1 - \sum_{k=1}^K \phi_k \right)$$

Taking the partial derivative with respect to $\phi_k$ and setting it to $0$:

$$\frac{\partial \mathcal{L}}{\partial \phi_k} = \sum_{i=1}^n \frac{\gamma_{ik}}{\phi_k} - \lambda = 0 \implies \phi_k = \frac{\sum_{i=1}^n \gamma_{ik}}{\lambda}$$

Summing both sides over $k$ and using $\sum_k \gamma_{ik} = 1$ and $\sum_k \phi_k = 1$ reveals that $\lambda = n$. Defining the effective number of points assigned to cluster $k$ as $N_k = \sum_{i=1}^n \gamma_{ik}$, we get:

$$\phi_k^{\text{new}} = \frac{N_k}{n}$$

### 2. Mean Vectors ($\mu_k$)
Isolating terms in $Q$ that depend on $\mu_k$ and expanding the log Gaussian density:

$$Q(\mu_k) = \sum_{i=1}^n \gamma_{ik} \left( -\frac{1}{2} (x_i - \mu_k)^T \Sigma_k^{-1} (x_i - \mu_k) \right) + \text{constant}$$

Taking the derivative with respect to $\mu_k$:

$$\nabla_{\mu_k} Q = \sum_{i=1}^n \gamma_{ik} \Sigma_k^{-1} (x_i - \mu_k) = 0$$

Multiplying by $\Sigma_k$ and expanding:

$$\sum_{i=1}^n \gamma_{ik} x_i - \left(\sum_{i=1}^n \gamma_{ik}\right) \mu_k = 0 \implies \mu_k^{\text{new}} = \frac{1}{N_k} \sum_{i=1}^n \gamma_{ik} x_i$$

### 3. Covariance Matrices ($\Sigma_k$)
Maximizing $Q$ with respect to the precision matrix $\Sigma_k^{-1}$ follows standard matrix calculus rules for Gaussian MLE:

$$\Sigma_k^{\text{new}} = \frac{1}{N_k} \sum_{i=1}^n \gamma_{ik} (x_i - \mu_k^{\text{new}})(x_i - \mu_k^{\text{new}})^T$$

### Role of Responsibilities
The responsibility $\gamma_{ik}$ acts as a continuous **fractional weight**. Instead of a data point contributing fully to just one cluster, its influence is split across all components. A point with high clarity ($\gamma_{ik} \approx 1.0$) strongly shapes the parameters of cluster $k$. Conversely, an ambiguous point falling between clusters distributes its weight smoothly, preventing minor variations from disproportionately pulling any single cluster's mean or covariance.

---

## Part 9: Structural Synthesis

Gaussian Mixture Model clustering can be viewed as an iterative process of conditional updates. The model coordinates prior beliefs and empirical evidence through the following structure:

*   The mixture weight $\phi_k$ represents the **unconditional prior probability** of an observation falling into cluster $k$.
*   The conditional density value $\mathscr{N}(x_i \mid \mu_k, \Sigma_k)$ acts as a likelihood score, measuring the **geometric compatibility** between the coordinates of $x_i$ and the shape of cluster $k$.
*   By blending these terms via Bayes' rule, the algorithm computes the responsibility $\gamma_{ik}$, which represents the **posterior probability** that data point $x_i$ belongs to cluster $k$.
*   These individual probabilities form the soft assignment vector $\mathbb{E}[Z_i \mid X_i = x_i]$.
*   During the M-step, the algorithm uses these posterior probabilities as weights to update the cluster parameters ($\phi_k, \mu_k, \Sigma_k$).

This confirms that Gaussian mixture clustering is fundamentally a **probabilistic clustering framework** driven by the conditional expectations of latent membership variables.

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
import plotly.express as px
import plotly.graph_objects as go

class GMMFinancialSegmenter:
    """
    An enterprise-grade segmenter that fits a Gaussian Mixture Model (GMM)
    on financial behaviors and provides robust interactive visualizations via Plotly.
    """
    def __init__(self, n_components=3, random_state=42):
        self.n_components = n_components
        self.random_state = random_state
        self.scaler = StandardScaler()
        self.model = GaussianMixture(
            n_components=self.n_components,
            covariance_type='full',
            random_state=self.random_state
        )
        self.X_train_scaled = None
        self.X_test_scaled = None
        self.X_train_raw = None
        self.X_test_raw = None
        self.feature_cols = None

    def load_and_preprocess(self, file_path='CC GENERAL.csv'):
        """Loads data, falls back to synthetic generation if missing, scales features, and splits sets."""
        if os.path.exists(file_path):
            print(f"[!] Target file '{file_path}' located. Extracting columns...")
            df = pd.read_csv(file_path)
        else:
            print(f"[!] '{file_path}' not found. Generating high-fidelity synthetic fallback...")
            np.random.seed(self.random_state)
            n_samples = 2000
            # Synthesize 3 overlapping financial archetypes
            c1 = np.random.multivariate_normal([800, 2000], [[40000, 20000], [20000, 90000]], int(n_samples*0.4))
            c2 = np.random.multivariate_normal([3500, 7500], [[150000, -30000], [-30000, 250000]], int(n_samples*0.35))
            c3 = np.random.multivariate_normal([150, 4500], [[8000, 4000], [4000, 120000]], int(n_samples*0.25))
            data = np.vstack([c1, c2, c3])
            df = pd.DataFrame(data, columns=['PURCHASES', 'CREDIT_LIMIT'])
            # Ensure no negative values are introduced by random normal simulation
            df = df.clip(lower=0)

        self.feature_cols = ['PURCHASES', 'CREDIT_LIMIT']
        X = df[self.feature_cols].dropna()

        # Perform 80/20 out-of-sample split
        X_train, X_test = train_test_split(X, test_size=0.20, random_state=self.random_state)
        self.X_train_raw = X_train.values
        self.X_test_raw = X_test.values

        # Scale features using training statistics to avoid data leakage
        self.X_train_scaled = self.scaler.fit_transform(X_train)
        self.X_test_scaled = self.scaler.transform(X_test)
        print(f" -> Data processing complete. Train shape: {self.X_train_scaled.shape}, Test shape: {self.X_test_scaled.shape}")

    def fit_em(self):
        """Executes the Expectation-Maximization algorithm on the scaled training matrix."""
        print("[*] Initiating EM Optimization Steps...")
        self.model.fit(self.X_train_scaled)
        print(f" -> Convergence Status: {self.model.converged_}")
        print(f" -> Total Iterations: {self.model.n_iter_}")

        train_ll = self.model.score(self.X_train_scaled)
        test_ll = self.model.score(self.X_test_scaled)
        print(f" -> Avg Log-Likelihood (Train): {train_ll:.4f}")
        print(f" -> Avg Log-Likelihood (Test Generalization): {test_ll:.4f}")

    def plot_empirical_density(self):
        """Generates an empirical 2D Density Heatmap with marginal histograms."""
        df_train = pd.DataFrame(self.X_train_raw, columns=self.feature_cols)
        fig = px.density_heatmap(
            df_train, x='PURCHASES', y='CREDIT_LIMIT',
            marginal_x='histogram', marginal_y='histogram',
            title='Figure 1: Empirical 2D Density Heatmap (Training Data)',
            labels={'PURCHASES': 'Purchases ($)', 'CREDIT_LIMIT': 'Credit Limit ($)'},
            color_continuous_scale=px.colors.sequential.Viridis
        )
        fig.update_layout(template='plotly_dark', width=900, height=650)
        fig.show()

    def _generate_contour_mesh(self, resolution=200):
        """Creates a coordinate mesh in raw space and evaluates max posterior responsibilities."""
        x_min, x_max = self.X_train_raw[:, 0].min() - 500, self.X_train_raw[:, 0].max() + 500
        y_min, y_max = self.X_train_raw[:, 1].min() - 500, self.X_train_raw[:, 1].max() + 500

        x_grid = np.linspace(x_min, x_max, resolution)
        y_grid = np.linspace(y_min, y_max, resolution)
        xx, yy = np.meshgrid(x_grid, y_grid)

        mesh_points_raw = np.c_[xx.ravel(), yy.ravel()]
        mesh_points_scaled = self.scaler.transform(mesh_points_raw)

        # Compute responsibilities for each coordinate point on the grid
        responsibilities = self.model.predict_proba(mesh_points_scaled) # Shape: (resolution^2, K)
        max_resp = np.max(responsibilities, axis=1).reshape(xx.shape)

        return x_grid, y_grid, max_resp

    def plot_assignment(self, mode='train'):
        """Overlays either training or testing data on top of the continuous responsibility fields."""
        x_grid, y_grid, max_resp = self._generate_contour_mesh()

        if mode == 'train':
            data_points = self.X_train_raw
            scaled_points = self.X_train_scaled
            title_text = "Figure 2: Training Assignments Over Maximum Posterior Responsibilities"
        else:
            data_points = self.X_test_raw
            scaled_points = self.X_test_scaled
            title_text = "Figure 3: Out-of-Sample Test Assignments Over Continuous Fields"

        # Determine hard assignments for the data points
        hard_labels = self.model.predict(scaled_points)

        fig = go.Figure()

        # Add the continuous background contour map of responsibilities
        fig.add_trace(go.Contour(
            x=x_grid, y=y_grid, z=max_resp,
            colorscale='Cividis', contours_coloring='heatmap',
            colorbar=dict(title="Max Posterior Prob $\\max_k \\gamma_{ik}$"),
            line_width=0, opacity=0.85, hoverinfo='skip'
        ))

        # Overlay the discrete customer scatter points
        for k in range(self.n_components):
            mask = (hard_labels == k)
            fig.add_trace(go.Scatter(
                x=data_points[mask, 0], y=data_points[mask, 1],
                mode='markers', name=f'Cluster {k+1}',
                marker=dict(size=5, line=dict(width=0.5, color='white'))
            ))

        fig.update_layout(
            title=title_text,
            xaxis_title="Purchases ($)", yaxis_title="Credit Limit ($)",
            template='plotly_dark', width=950, height=650,
            legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
        )
        fig.show()

# Execution Pipeline
segmenter = GMMFinancialSegmenter(n_components=3)
segmenter.load_and_preprocess('CC GENERAL.csv')
segmenter.fit_em()

# Render Interactive Plots
segmenter.plot_empirical_density()
segmenter.plot_assignment(mode='train')
segmenter.plot_assignment(mode='test')
```

## Evaluation of Results and Mathematical Alignment

### 1. Empirical 2D Density Heatmap (Figure 1)
Figure 1 shows the distribution of the raw training data. The marginal histograms reveal clear asymmetry and multimodality along both axes. Standard linear clustering algorithms (like K-Means) assume spherical distributions and struggle with this structure. The GMM accommodates these characteristics by assigning a full covariance matrix $\Sigma_k$ to each component.

### 2. Generalization and Test Log-Likelihood Validation
The model outputs the average log-likelihood for both the training and unseen test sets. A small gap between the training log-likelihood and the test log-likelihood indicates good generalization, showing that the learned density functions fit the underlying distribution well without overfitting.

### 3. Continuous Background Contours and Analytical Verification (Figures 2 & 3)
The continuous background contour map shows how the model handles uncertainty across the feature space.

In Part 3, we proved that the soft cluster assignment vector is the conditional expectation of the one-hot encoded latent variable:

$$\mathbb{E}[Z_i \mid X_i = x_{\text{grid}}] = \begin{bmatrix} \gamma_{i1} \\ \gamma_{i2} \\ \vdots \\ \gamma_{iK} \end{bmatrix}$$

The background map plots the maximum value in this vector ($\max_k \gamma_{ik}$) at every point on the grid:
*   **High Certainty Regions (Yellow/Bright Zones):** In the dense centers of each cluster, the maximum posterior probability approaches $1.0$. Here, the conditional expectation vector points strongly toward a single vertex in the probability simplex.
*   **Ambiguity Regions (Dark/Transition Lines):** Where the clusters overlap, the maximum responsibility drops toward $1.0 / K \approx 0.33$. The contour map turns dark in these areas, visually highlighting regions of uncertainty. This shows how the soft assignment vector changes continuously as you move between cluster centers, directly illustrating the math derived in Part 3.
