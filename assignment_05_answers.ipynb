\section*{Question 1: Bias of the Covariance MLE and Bessel's Correction}

The raw Maximum Likelihood Estimator (MLE) for the population covariance matrix $\boldsymbol{\Sigma}$ is given by:
\begin{equation}
\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T
\end{equation}

\subsection*{1.1 Derivation of the Expectation of $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$}

To find $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}]$, we first rewrite the term inside the summation. Notice that the sample mean can be expanded as:
\begin{equation}
\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})
\end{equation}

Substituting this into the outer product yields:
\begin{align}
(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = {} & \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T \\
= {} & (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \nonumber \\
& - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T
\end{align}

Summing this expression over all $i = 1, 2, \dots, n$:
\begin{align}
\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = {} & \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \left[ \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right] (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \nonumber \\
& - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \left[ \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) \right]^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T
\end{align}

Recognizing that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$, the middle cross-terms simplify:
\begin{align}
\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = {} & \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \nonumber \\
& - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \\
= {} & \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T
\end{align}

Taking the expectation on both sides:
\begin{equation}
\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \sum_{i=1}^n \mathbb{E}\left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] - n\,\mathbb{E}\left[ (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right]
\end{equation}

By definition, for i.i.d. random vectors:
\begin{equation}
\mathbb{E}\left[ (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T \right] = \boldsymbol{\Sigma}
\end{equation}
\begin{equation}
\mathbb{E}\left[ (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T \right] = \mathrm{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}
\end{equation}

Substituting these components back into the expectation equation yields:
\begin{align}
\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] &= \sum_{i=1}^n \boldsymbol{\Sigma} - n \left( \frac{1}{n}\boldsymbol{\Sigma} \right) \\
&= n\boldsymbol{\Sigma} - \boldsymbol{\Sigma} = (n-1)\boldsymbol{\Sigma}
\end{align}

Dividing by the scalar $n$ to match the MLE definition demonstrates the inherent bias:
\begin{equation}
\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \frac{n-1}{n}\boldsymbol{\Sigma}
\end{equation}

\subsection*{1.2 Mathematical Proof of the Unbiased Estimator $\mathbf{S}$}

To construct an unbiased estimator, we introduce a scaling correction factor $c$ such that $\mathbf{S} = c \widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ satisfies $\mathbb{E}[\mathbf{S}] = \boldsymbol{\Sigma}$:
\begin{equation}
\mathbb{E}[\mathbf{S}] = c \,\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = c \left( \frac{n-1}{n} \right) \boldsymbol{\Sigma} = \boldsymbol{\Sigma} \implies c = \frac{n}{n-1}
\end{equation}

Applying Bessel's correction factor $c$ directly eliminates the bias:
\begin{equation}
\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left[ \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T
\end{equation}
This confirms that $\mathbf{S}$ is the unique, unbiased sample covariance estimator.

\section*{Question 2: Structural Health Monitoring and Decision Risk}

\subsection*{2.1 Operational Definitions of Error Frameworks}

In continuous structural health monitoring (SHM), the statistical testing framework is defined as:
\begin{itemize}
    \item $H_0$: The structural asset is healthy and undamaged (normal operation).
    \item $H_1$: The structural asset has sustained damage or degradation (anomalous state).
\end{itemize}

\begin{enumerate}
    \item \textbf{Type I Error ($\alpha$): False Alarm Rate} \\
    Occurs when the monitoring framework rejects $H_0$ given that $H_0$ is true. Mathematically:
    \begin{equation}
    \alpha = P(\text{Reject } H_0 \mid H_0 \text{ is true})
    \end{equation}
    In this application, it represents the probability that the system flags a healthy component as damaged, triggering unneeded maintenance down-time or an emergency plant shutdown.

    \item \textbf{Type II Error ($\beta$): Missed Detection Rate} \\
    Occurs when the framework fails to reject $H_0$ given that $H_1$ is true. Mathematically:
    \begin{equation}
    \beta = P(\text{Fail to Reject } H_0 \mid H_1 \text{ is true})
    \end{equation}
    In this application, it represents the probability that structural damage exists, but the monitoring layer fails to detect it, risking catastrophic asset failure.
\end{enumerate}

\subsection*{2.2 Consequence of an Ultra-Conservative Threshold ($\alpha = 0.0001$)}

\subsubsection*{Impact on Statistical Power}
Setting $\alpha = 0.0001$ shrinks the critical region (the rejection zone) of the null hypothesis distribution. Because $\alpha$ and $\beta$ are inversely linked for a fixed sample size $n$, minimizing the false alarm probability shifts the decision boundary far out into the tails of the $H_0$ distribution.

This directly inflates the probability of committing a Type II error ($\beta$). The statistical power of the diagnostic system, defined as:
\begin{equation}
\text{Power} = 1 - \beta = P(\text{Reject } H_0 \mid H_1 \text{ is true})
\end{equation}
will drop significantly. The diagnostic layer becomes insensitive to low-to-moderate structural anomalies, failing to trigger warnings unless damage is massively advanced.

\subsubsection*{Geometric Modification of the Confidence Ellipsoid}
In multivariate structural health monitoring, the threshold for healthy operation is defined geometrically by a Mahalanobis distance confidence ellipsoid centered at the sample mean vector $\widehat{\boldsymbol{\mu}}_n$:
\begin{equation}
(\mathbf{X} - \widehat{\boldsymbol{\mu}}_n)^T \mathbf{S}^{-1} (\mathbf{X} - \widehat{\boldsymbol{\mu}}_n) \le \chi^2_{m, 1-\alpha}
\end{equation}

Where $\chi^2_{m, 1-\alpha}$ is the critical value of a Chi-Squared distribution with $m$ degrees of freedom at a confidence level of $1-\alpha$.

When the engineer sets an ultra-conservative significance level $\alpha = 0.0001$, the target cumulative probability value shifts from a standard level (e.g., $1 - 0.05 = 0.95$) to $1 - 0.0001 = 0.9999$. This substantially increases the critical boundary value:
\begin{equation}
\chi^2_{m, 0.9999} \gg \chi^2_{m, 0.95}
\end{equation}

The volume $V$ of an $m$-dimensional hyper-ellipsoid is directly proportional to the critical bounding radius raised to the power of $m$:
\begin{equation}
V \propto \left(\sqrt{\chi^2_{m, 1-\alpha}}\right)^m = \left(\chi^2_{m, 1-\alpha}\right)^{m/2}
\end{equation}

Consequently, enforcing $\alpha = 0.0001$ \textbf{expands the geometric volume} of the "healthy operation" confidence ellipsoid dramatically. The operating window becomes so large that anomalous data vectors reflecting real degradation remain trapped inside the expanded healthy domain, leaving structural shifts hidden from detection.

\section*{Operationalizing the Multivariate Central Limit Theorem via Slutsky's Theorem}

\subsection*{1. Rigorous Statement of Slutsky's Theorem}

Let $\{\mathbf{Y}_n\}$ be a sequence of random vectors in $\mathbb{R}^m$, and let $\{\mathbf{Z}_n\}$ be a sequence of random matrices in $\mathbb{R}^{m \times m}$, defined on the same probability space.

If $\mathbf{Y}_n$ converges in distribution to a random vector $\mathbf{Y}$ as $n \to \infty$:
\begin{equation}
\mathbf{Y}_n \xrightarrow{d} \mathbf{Y}
\end{equation}
and $\mathbf{Z}_n$ converges in probability to a constant deterministic matrix $\mathbf{C}$ as $n \to \infty$:
\begin{equation}
\mathbf{Z}_n \xrightarrow{p} \mathbf{C}
\end{equation}

Then Slutsky's Theorem asserts the following joint asymptotic behaviors:
\begin{enumerate}
    \item \textbf{Additive Form:}
    \begin{equation}
    \mathbf{Y}_n + \mathbf{Z}_n \mathbf{v} \xrightarrow{d} \mathbf{Y} + \mathbf{C}\mathbf{v} \quad \text{for any constant vector } \mathbf{v} \in \mathbb{R}^m
    \end{equation}
    \item \textbf{Multiplicative Form:}
    \begin{equation}
    \mathbf{Z}_n \mathbf{Y}_n \xrightarrow{d} \mathbf{C}\mathbf{Y}
    \end{equation}
    \item \textbf{Inverse Multiplicative Form:} If $\mathbf{C}$ is non-singular and invertible, then:
    \begin{equation}
    \mathbf{Z}_n^{-1} \mathbf{Y}_n \xrightarrow{d} \mathbf{C}^{-1}\mathbf{Y}
    \end{equation}
\end{enumerate}

\subsection*{2. Proof of Substitution Justification via Slutsky's Theorem}

We aim to justify substituting the empirical sample covariance matrix $\mathbf{S}$ for the unobservable population covariance matrix $\boldsymbol{\Sigma}$ without altering the limiting distribution of our standardized statistic.

\subsubsection*{Step 1: Establishing Convergence Components}
From the Lindeberg-Lévy Multivariate Central Limit Theorem (CLT), we have our sequence of random vectors converging in distribution:
\begin{equation}
\mathbf{Y}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathbf{Y} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})
\end{equation}

By the Weak Law of Large Numbers (WLLN), the empirical sample covariance matrix $\mathbf{S}$ is a consistent estimator for the true population parameter $\boldsymbol{\Sigma}$. Since matrix inversion is a continuous mapping for non-singular matrices, applying the Continuous Mapping Theorem yields the following matrix convergence in probability:
\begin{equation}
\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \implies \mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}
\end{equation}
Here, $\mathbf{S}^{-1/2}$ represents our sequence of random matrices $\mathbf{Z}_n$, and $\boldsymbol{\Sigma}^{-1/2}$ acts as the constant matrix $\mathbf{C}$.

\subsubsection*{Step 2: Applying Slutsky's Theorem}
We construct a studentized Wald-type distance vector by combining our matrix sequence and vector sequence. By the Multiplicative Form of Slutsky's Theorem ($\mathbf{Z}_n \mathbf{Y}_n \xrightarrow{d} \mathbf{C}\mathbf{Y}$):
\begin{equation}
\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathbf{Y}
\end{equation}

Because $\mathbf{Y}$ follows a zero-mean normal distribution with a covariance structure equal to $\boldsymbol{\Sigma}$, the linear transformation on the right-hand side simplifies to the standard multivariate normal distribution:
\begin{equation}
\boldsymbol{\Sigma}^{-1/2} \mathbf{Y} \sim \mathscr{N}\left( \mathbf{0}, \, \boldsymbol{\Sigma}^{-1/2} \boldsymbol{\Sigma} (\boldsymbol{\Sigma}^{-1/2})^T \right) = \mathscr{N}(\mathbf{0}, \mathbf{I}_m)
\end{equation}

Therefore, substituting $\mathbf{S}$ preserves the standard normal convergence limit:
\begin{equation}
\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I}_m)
\end{equation}

\subsubsection*{Step 3: Operationalizing the Dashboard Representation}
For large finite samples ($n \gg 1$), we can map this asymptotic distribution back into an unstandardized, operational coordinate framework by scaling out the transformation.

Multiplying back by the estimated scaling factor $\frac{1}{\sqrt{n}}\mathbf{S}^{1/2}$ maps the distribution to:
\begin{equation}
(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \approx \mathscr{N}\left(\mathbf{0}, \frac{1}{n}\mathbf{S}\right)
\end{equation}

Treating the empirical sample observations as fixed anchor points within our real-time monitoring dashboard allows us to model the distribution of the true mean vector parametrically around our current sample point:
\begin{equation}
\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\widehat{\boldsymbol{\mu}}_n, \frac{1}{n}\mathbf{S}\right)
\end{equation}
This confirms that the error introduced by swapping $\boldsymbol{\Sigma}$ with $\mathbf{S}$ decays to zero as $n \to \infty$, validating the plug-in step mathematically.

Let $\widetilde{\mathbf{X}}_i = (\mathbf{X}_i - \boldsymbol{\mu}) \in \mathbb{R}^m$ represent a zero-mean, centered baseline feature vector. Its population covariance matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{m \times m}$ is defined by:
\begin{equation}
\boldsymbol{\Sigma} = \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T]
\end{equation}
By the spectral theorem for symmetric, positive-semidefinite matrices, we decompose the covariance matrix as:
\begin{equation}
\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T
\end{equation}
where $\mathbf{P} \in \mathbb{R}^{m \times m}$ is an orthogonal matrix whose columns represent the eigenvectors of $\boldsymbol{\Sigma}$, satisfying $\mathbf{P}^T\mathbf{P} = \mathbf{P}\mathbf{P}^T = \mathbf{I}_m$. The matrix $\mathbf{\Lambda}$ is a diagonal matrix containing sorted eigenvalues:
\begin{equation}
\mathbf{\Lambda} = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_m), \quad \lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_m \ge 0
\end{equation}

We define the linear, orthogonal projection mapping into the principal space as:
\begin{equation}
\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i
\end{equation}

\subsection{Step-by-Step Proof of Covariance Diagonalization}
To find the theoretical covariance matrix of the transformed coordinate vector, we compute the expectation of its outer product, $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$:
\begin{align}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] &= \mathbb{E}\left[ \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right) \left(\mathbf{P}^T \widetilde{\mathbf{X}}_i\right)^T \right] \\
&= \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \left(\widetilde{\mathbf{X}}_i^T \mathbf{P}\right) \right] \\
&= \mathbb{E}\left[ \mathbf{P}^T \left( \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \right) \mathbf{P} \right]
\end{align}

Since the projection matrix $\mathbf{P}$ is deterministic, it remains invariant under the expectation operator. By leveraging the linearity of expectation, we pull $\mathbf{P}^T$ and $\mathbf{P}$ outside the operator:
\begin{equation}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \, \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \, \mathbf{P}
\end{equation}

Substituting the definition of the baseline population covariance matrix $\boldsymbol{\Sigma}$:
\begin{equation}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}
\end{equation}

Next, we substitute the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ into the expression:
\begin{equation}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \left( \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T \right) \mathbf{P}
\end{equation}

Using the associativity property of matrix multiplication, we regroup the terms:
\begin{equation}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \left(\mathbf{P}^T \mathbf{P}\right) \mathbf{\Lambda} \left(\mathbf{P}^T \mathbf{P}\right)
\end{equation}

Because $\mathbf{P}$ is an orthonormal basis matrix, $\mathbf{P}^T \mathbf{P} = \mathbf{I}_m$. Substituting the identity matrix simplifies the expression exactly to the diagonal eigenvalue matrix:
\begin{equation}
\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I}_m \, \mathbf{\Lambda} \, \mathbf{I}_m = \mathbf{\Lambda}
\end{equation}

\subsection{Geometric and Statistical Interpretation}
\begin{itemize}
    \item \textbf{Statistical Interpretation:} Because $\mathbf{\Lambda}$ contains zero values everywhere outside its main diagonal, the cross-covariance between any two distinct transformed coordinates $Z_{i,j}$ and $Z_{i,k}$ is exactly zero for $j \neq k$:
    \begin{equation}
    \text{Cov}(Z_{i,j}, Z_{i,k}) = 0
    \end{equation}
    This proves that the orthogonal projection completely de-correlates the underlying features, isolating independent processes from multi-channel sensor noise.
    \item \textbf{Geometric Interpretation:} The transformation represents a rigid rotation of the original coordinate frames. The rows of $\mathbf{P}^T$ form a new Cartesian coordinate system whose axes align perfectly with the hyper-ellipsoidal axes of data dispersion, preventing geometric skew.
\end{itemize}

\subsection{Formal Proof of Total Variance Invariance}
The total variance of a multivariate system is defined as the sum of variances along its individual coordinate axes, which mathematically matches the trace ($\text{tr}$) of its covariance matrix.

We invoke the cyclic permutation invariance property of the matrix trace, which states that for any matrices $\mathbf{A}$, $\mathbf{B}$, and $\mathbf{C}$ of matching dimensions, $\text{tr}(\mathbf{A}\mathbf{B}\mathbf{C}) = \text{tr}(\mathbf{C}\mathbf{A}\mathbf{B})$. Setting $\mathbf{A} = \mathbf{P}$, $\mathbf{B} = \mathbf{\Lambda}$, and $\mathbf{C} = \mathbf{P}^T$:
\begin{equation}
\text{tr}(\boldsymbol{\Sigma}) = \text{tr}\left( \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T \right)
\end{equation}

Permuting the trailing matrix $\mathbf{P}^T$ to the front of the trace argument yields:
\begin{equation}
\text{tr}(\boldsymbol{\Sigma}) = \text{tr}\left( \mathbf{P}^T \mathbf{P} \mathbf{\Lambda} \right)
\end{equation}

Since the transformation matrix is orthogonal, we substitute $\mathbf{P}^T \mathbf{P} = \mathbf{I}_m$:
\begin{equation}
\text{tr}(\boldsymbol{\Sigma}) = \text{tr}\left( \mathbf{I}_m \mathbf{\Lambda} \right) = \text{tr}(\mathbf{\Lambda})
\end{equation}

Expanding the trace of the diagonalized matrix confirms that total system energy is conserved:
\begin{equation}
\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j
\end{equation}

\subsection{Rigorous Subspace Ratio Formulations}
When dealing with an empirical sample covariance matrix $\mathbf{S}$ and its sorted sample eigenvalues $\widehat{\lambda}_1 \ge \widehat{\lambda}_2 \ge \dots \ge \widehat{\lambda}_m$, we partition the system into an explained subspace ($\widehat{\mathbf{P}}_k \in \mathbb{R}^{m \times k}$) and an unexplained residual subspace ($\widehat{\mathbf{P}}_{m-k} \in \mathbb{R}^{m \times (m-k)}$).

The tracking parameters for evaluating subspace truncation are formulated as:
\begin{enumerate}
    \item \textbf{Cumulative Explained Variance Ratio $\Phi(k)$:}
    \begin{equation}
    \Phi(k) = \frac{\text{tr}\left( \widehat{\mathbf{P}}_k^T \mathbf{S} \widehat{\mathbf{P}}_k \right)}{\text{tr}(\mathbf{S})} = \frac{\sum_{j=1}^k \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j}
    \end{equation}
    \item \textbf{Residual Unexplained Variance Ratio $\Psi(k)$:}
    \begin{equation}
    \Psi(k) = \frac{\text{tr}\left( \widehat{\mathbf{P}}_{m-k}^T \mathbf{S} \widehat{\mathbf{P}}_{m-k} \right)}{\text{tr}(\mathbf{S})} = \frac{\sum_{j=k+1}^m \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j} = 1 - \Phi(k)
    \end{equation}
\end{enumerate}

\section{Subspace Energy Decomposition and Real-Time Diagnostics}
% =========================================================================

In real-time structural health monitoring, the centered raw snapshot vector $(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) \in \mathbb{R}^m$ is projected onto the orthogonal subspaces:
\begin{equation}
(\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}
\end{equation}
where $\mathbf{z}_{i,k} \in \mathbb{R}^k$ represents the dominant scores, and $\mathbf{z}_{i,m-k} \in \mathbb{R}^{m-k}$ represents the residual components.

We define the reconstructed profile vector $\widehat{\mathbf{x}}_i$ and the residual error vector $\mathbf{e}_i$ as:
\begin{equation}
\widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} \quad \text{and} \quad \mathbf{e}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}
\end{equation}

\subsection{Proof of the Subspace Pythagorean Identity}
We derive the squared Euclidean norm of the centered observation vector by expressing it as an inner vector product:
\begin{align}
\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 &= (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n)^T (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) \\
&= \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)
\end{align}

Expanding this product yields four distinct components:
\begin{align}
\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = {} & \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \nonumber \\
& + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}
\end{align}

Because the primary matrix block $\widehat{\mathbf{P}}_k$ and the residual matrix block $\widehat{\mathbf{P}}_{m-k}$ represent mutually orthogonal eigenspaces, their cross-products collapse to zero blocks:
\begin{equation}
\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}_{k \times (m-k)} \quad \text{and} \quad \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k = \mathbf{0}_{(m-k) \times k}
\end{equation}
Furthermore, since each matrix contains orthonormal individual columns, we have $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_k$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{m-k}$.

Substituting these simplifications back into the norm expansion removes the cross-subspace components:
\begin{align}
\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 &= \mathbf{z}_{i,k}^T \left( \mathbf{I}_k \right) \mathbf{z}_{i,k} + \mathbf{0} + \mathbf{0} + \mathbf{z}_{i,m-k}^T \left( \mathbf{I}_{m-k} \right) \mathbf{z}_{i,m-k} \\
&= \mathbf{z}_{i,k}^T \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k} \\
&= \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2
\end{align}
This proves that total raw observation energy decomposes into orthogonal, independent subspace tracking layers.

\subsection{Diagnostic Allocation and Sensitivity Analysis}
To monitor anomalies, we implement two complementary statistical tracking metrics:
\begin{equation}
T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j} \quad \text{and} \quad Q_i = \|\mathbf{e}_i\|^2 = \sum_{j=k+1}^m z_{i,j}^2
\end{equation}

\begin{enumerate}
    \item \textbf{Environmental Load Anomaly (e.g., Extreme Wind) $\longrightarrow$ Tracked via $T^2$:} \\
    \textit{Physical Justification:} Extreme wind forces massive excitation along existing structural modes without changing the underlying physical boundary constraints. The structure behaves normally but experiences out-of-distribution amplitudes. Because $T^2$ scales coordinates by their inverse eigenvalues ($\widehat{\lambda}_j^{-1}$), it acts as a mahalanobis-type distance filter that is highly sensitive to extreme movements within the nominal operational space.
    \item \textbf{Internal Structural Fracture (e.g., Fatigue Crack) $\longrightarrow$ Tracked via $Q$ (SPE):} \\
    \textit{Physical Justification:} An internal fatigue crack changes the structural stiffness tensor, breaking the nominal correlations among sensors. This physical state change forces the data vector to project in new directions that are orthogonal to the normal modes. This unmodeled energy projects directly onto the residual subspace, causing a sharp spike in the $Q$ statistic.
\end{enumerate}

class SubspaceDiagnostics:
    def __init__(self, n_sensors=4):
        """
        Initializes the multivariate structural diagnostics layer.
        """
        self.m = n_sensors
        self.mu = None
        self.P = None
        self.lambdas = None

    def analyze_truncation_boundaries(self, X_raw):
        """
        Calculates Hotelling's T2 and residual Q statistics across multiple cutoffs.
        
        Parameters:
        -----------
        X_raw : np.ndarray
            Shape (n, m) - Matrix containing raw sensor snapshots.
            
        Returns:
        --------
        metrics_profile : dict
            Empirical expectations of T2 and Q mapped across k in {1, 2, 3}.
        """
        n, m = X_raw.shape
        assert m == self.m, f"Dimension error. Expected {self.m} channels, got {m}."
        
        # 1. Center and Standardize the empirical baseline data
        self.mu = np.mean(X_raw, axis=0)
        X_centered = X_raw - self.mu
        
        # Compute the unbiased sample covariance matrix S
        S = (X_centered.T @ X_centered) / (n - 1)
        
        # 2. Decompose S via Hermitian Eigh and sort in descending order
        eigenvalues, eigenvectors = np.linalg.eigh(S)
        
        idx_descending = np.argsort(eigenvalues)[::-1]
        self.lambdas = eigenvalues[idx_descending]
        self.P = eigenvectors[:, idx_descending]
        
        # Transform centered observations into rotated Principal Component scores
        Z = X_centered @ self.P  # Shape: (n, m)
        
        metrics_profile = {}
        
        # 3. Loop through every possible subspace cutoff value k in {1, 2, 3}
        for k in [1, 2, 3]:
            # Partition scores into principal and residual parts
            Z_k = Z[:, :k]
            Z_residual = Z[:, k:]
            
            lambdas_k = self.lambdas[:k]
            
            # Hotelling's T2 array calculation via vectorized broadcasting
            T2_array = np.sum((Z_k ** 2) / lambdas_k, axis=1)
            
            # Q (SPE) residual array calculation
            Q_array = np.sum(Z_residual ** 2, axis=1)
            
            # 4. Compute empirical expectations (means) across all samples
            expected_T2 = np.mean(T2_array)
            expected_Q = np.mean(Q_array)
            
            metrics_profile[k] = {
                "expected_T2": expected_T2,
                "expected_Q": expected_Q
            }
            
        return metrics_profile

def run_factor_analysis_pipeline(df, k_factors):
    """
    Standardizes input observations, runs a parametric Factor Analysis model
    with Varimax rotation, and extracts latent states via Thomson's method.
    """
    # Rule 1: Isolate numeric observation columns
    numeric_df = df.select_dtypes(include=[np.number])
    feature_names = list(numeric_df.columns)
    X = numeric_df.to_numpy()
    n, m = X.shape
    
    # Rule 2: Validation exceptions and dimensionality constraints
    if n <= m:
        raise ValueError(f"Processing Rejected: Snapshot count n={n} must exceed channel count m={m}.")
    if k_factors >= m:
        raise ValueError(f"Dimensionality Error: Latent factors k={k_factors} must be strictly less than features m={m}.")
        
    # Rule 3: Z-Score mapping layer with explicit variance guardrails
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0, ddof=1)
    
    # Cap near-zero variance channels to avoid runtime divide-by-zero exceptions
    sigma = np.where(sigma < 1e-15, 1e-15, sigma)
    Z = (X - mu) / sigma
    
    # Rule 4: Information Decomposition Layer (FA + Varimax)
    # sklearn applies Varimax via an internal SVD/maximum likelihood optimization mapping
    fa = FactorAnalysis(n_components=k_factors, rotation='varimax', random_state=42)
    fa.fit(Z)
    
    # Extract structural component parameters
    lambda_rotated = fa.components_.T  # Shape: (m, k)
    uniqueness = fa.noise_variance_     # Shape: (m,)
    communality = 1.0 - uniqueness     # Shape: (m,) Since R_jj = 1 for standardized data
    
    # Rule 5: Mathematical State Reconstruction (Thomson's MMSE method)
    # For standardized data, R ≈ Lambda @ Lambda.T + Psi

    R_inv = np.linalg.inv(R_implied)
    
    # Thomson score projection matrix mapping equation: F = Z @ R^-1 @ Lambda
    score_weights = R_inv @ lambda_rotated
    F = Z @ score_weights              # Shape: (n, k)
    
    # Rule 6: Performance Diagnostic Telemetry
    factor_variances = np.var(F, axis=0, ddof=1)
    avg_communality = np.mean(communality) * 100

\subsection*{1.1 Conceptual Contrast: Raw PCA vs. Varimax-Rotated Loading Vectors}
\begin{itemize}
    \item \textbf{Raw Principal Component Loadings:} These vectors are derived strictly to maximize the global variance explained by each consecutive component. Geometrically, the first loading vector aligns with the direction of maximum dispersion in the data space. Consequently, raw PCA loading vectors are often unconstrained, meaning sensors load broadly across multiple components, which creates a dense matrix that makes it difficult to isolate distinct physical processes.
    \item \textbf{Varimax-Rotated Factor Loadings:} The Varimax criterion performs an orthogonal rotation that maximizes the variance of the squared loadings within each column of the factor loading matrix $\mathbf{\Lambda}$. Mathematically, it maximizes the objective function:
    \begin{equation}
    V = \sum_{l=1}^k \left[ \frac{1}{m} \sum_{j=1}^m \lambda_{jl}^4 - \left( \frac{1}{m} \sum_{j=1}^m \lambda_{jl}^2 \right)^2 \right]
    \end{equation}
    This pushes the factor loading coefficients toward $1$ or $0$. Conceptually, this yields a "simple structure" where each sensor loads heavily on only one (or a very few) latent factors, allowing an engineer to directly match specific factors to isolated physical phenomena (e.g., separating thermal cycles from mechanical strain vibration).
\end{itemize}

It is completely decoupled from the real physical processes ($f_1, f_2$) driving the plant asset. Its data profile consists entirely of isolated white noise or severe electrical fault energy.
* **Part 2 (The PCA Loading Contamination):** Looking at the **Feature Loading Matrix** in the PCA dashboard, we see a striking anomaly: `PC 1` maps almost exclusively to a single variable—`Sensor_4` (indicated by the deep red block with a loading value near $1.0$). Because `Sensor_4` has a massive, unmitigated noise variance scale ($\sigma^2 \approx 2.0$), the PCA engine maximizes global variance by dedicating its entire primary axis (`PC 1`) to tracking this single noisy sensor. It completely isolates `Sensor_1`, `Sensor_2`, and `Sensor_3` to a loading of $0$ on `PC 1`.
* **Part 3 (Engineering Risk):** This presents a significant operational hazard. If you use standard PCA for monitoring, a single blown or short-circuited instrument will hijack your most critical monitoring axis (`PC 1`), inflating its eigenvalue to $\lambda_1 \approx 2.0$. This masks the true common underlying structural dynamics of the asset and renders your primary control chart useless for catching actual systemic failures.

### Solution for Question 2: Decoupling Structural Loading via Rotation

* **Part 1 (Mathematical Jumble vs. Simple Structure):** Because PCA is bound to strict, orthogonal variance maximization, look at what happens to the remaining signals on `PC 2` and `PC 4`. `PC 2` and `PC 4` exhibit high loading values across *both* `Sensor_1` and `Sensor_2` simultaneously, while also bleeding into `Sensor_3`. This cross-loading represents a mathematically forced mixture of signals. Factor Analysis applies a **Varimax rotation** to eliminate this. By maximizing loading variance within columns, it completely isolates the signals: `Sensor_1` and `Sensor_2` group cleanly into `Factor 1`, while `Sensor_3` shifts entirely into `Factor 2`.
* **Part 2 (Operational Interpretability):** The PCA loading matrix forces a complex interpretation. If a diagnostic alarm trips on `PC 2`, an engineer has to figure out a messy combination of Sensors 1, 2, and 3. With the rotated FA framework, the layout is intuitive. `Factor 1` represents the isolated physical process causing redundancy in `Sensor_1` and `Sensor_2`, while `Factor 2` maps directly to the independent process monitored by `Sensor_3`. Troubleshooting root causes becomes instantaneous.


### Solution for Question 3: Determining Subspace Truncation ($k$)

* **Part 1 (Curve Observations):** Moving from $k=1$ to $k=2$, the Mean $Q$ Statistic (SPE) plunges from $2.4$ down to $0.9$. When transitioning from $k=2$ to $k=3$, the curve flattens out into an explicit "elbow", decreasing at a much slower, linear rate down toward $0.1$.
* **Part 2 (The True Intrinsic Dimension):** Cross-referencing the $Q$ curve with the new loading matrix makes the selection clear. `PC 1` tracks the isolated noise of `Sensor_4`. `PC 2` captures the true, shared variance of the working sensors (`Sensor_1`, `Sensor_2`, `Sensor_3`). The sharp drop in the $Q$ statistic at $k=2$ tells us that once these two distinct behaviors are accounted for, the structured information is fully exhausted. Thus, $k=2$ represents the true hidden physical dimensionality of the asset.
* **Part 3 (The Danger of Over-fitting to $k=3$):** Look at `PC 3` in the loading matrix—it maps exclusively to the leftover residual variance of `Sensor_3`. Setting your truncation limit to $k=3$ pulls this residual space into your "clean" model subspace. This overfits the system, causing your Hotelling's $T^2$ control limits to trigger frequent false alarms due to minor, normal baseline variations.



### Solution for Question 4: Operational Trade-offs in System Health Monitoring

* **The PCA Strategy:** Tracking the main principal subspace via $T^2$ alongside the residual space via the $Q$ statistic.
* *Vulnerability:* This approach is highly vulnerable to sensor degradation. As shown in the PCA loading matrix, if a single channel (`Sensor_4`) fails high, it structurally shifts the orientation of your principal axes, dragging the calculations for the remaining healthy channels along with it.


* **The FA Strategy:** Directly monitoring the rotated Latent Factor Scores.
* *Robustness:* This approach is highly robust. Because Factor Analysis explicitly separates localized noise via the diagonal uniqueness vector ($\Psi$), any single-channel sensor failure is safely absorbed by that channel's uniqueness profile ($\psi_4^2 \rightarrow 1$). The underlying latent factors (`Factor 1` and `Factor 2`) remain clean, stable, and uncorrupted, allowing the rest of your health monitoring network to function normally.
