# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

Answer

# 1. The Partition Function

The canonical partition function $Z(T)$ is given by the sum over all quantum states:

$$Z(T) = \sum_{j=0}^\infty e^{-\frac{E_j}{k_B T}} = \sum_{j=0}^\infty e^{-\frac{\hbar \omega \left(j + \frac{1}{2}\right)}{k_B T}}$$

We can factor out the constant ground-state energy term:

$$Z(T) = e^{-\frac{\hbar \omega}{2k_B T}} \sum_{j=0}^\infty \left(e^{-\frac{\hbar \omega}{k_B T}}\right)^j$$

The summation is an infinite geometric series of the form $\sum_{j=0}^\infty x^j$, where $x = e^{-\frac{\hbar \omega}{k_B T}}$. Because $\hbar, \omega, k_B, T > 0$, the ratio satisfies $0 < x < 1$, which guarantees that the series strictly converges.

Using the geometric series convergence formula $\sum_{j=0}^\infty x^j = \frac{1}{1-x}$, we get:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

---

# 2. Thermodynamic Potentials

### Helmholtz Free Energy, $F(T)$
The Helmholtz free energy is defined as $F(T) = -k_B T \ln Z$. Substituting our expression for $Z(T)$:

$$F(T) = -k_B T \ln \left( \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \right)$$

Applying logarithmic identities:

$$F(T) = -k_B T \left[ \ln\left(e^{-\frac{\hbar \omega}{2k_B T}}\right) - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

### Internal Energy, $U(T)$
To find the internal energy using $\beta = \frac{1}{k_B T}$, we first rewrite $\ln Z$ in terms of $\beta$:

$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$

Taking the negative derivative with respect to $\beta$:

$$U(T) = -\frac{\partial \ln Z}{\partial \beta} = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot (-\hbar \omega) \right]$$

$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

Multiplying the numerator and denominator by $e^{\beta \hbar \omega}$ yields the standard form:

$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1} = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

---

# 3. Entropy

We find the entropy $S(T)$ via the relation $S = -\left(\frac{\partial F}{\partial T}\right)$:

$$S = -\frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

Applying the product rule:

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar \omega}{k_B T}}\right) \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right)$$

Simplifying the terms:

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \frac{e^{-\frac{\hbar \omega}{k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)}$$

### Verification of $U = F + TS$
Let's construct $F + TS$:

$$F + TS = \left[\frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)\right] + T \left[-k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)}\right]$$

The logarithmic terms cancel out exactly:

$$F + TS = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U(T)$$

The relation holds perfectly.

---

# 4. Heat Capacity and Hyperbolic Form

The constant-volume heat capacity is given by $C_V = \frac{\partial U}{\partial T}$:

$$C_V = \frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right]$$

Using the chain rule:

$$C_V = \hbar \omega \cdot \left[ -\frac{1}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} \right] \cdot e^{\frac{\hbar \omega}{k_B T}} \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right)$$

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

To convert this to hyperbolic form, we divide the top and bottom by $e^{\frac{\hbar \omega}{k_B T}}$ (which inside the squared denominator means dividing the inside by $e^{\frac{\hbar \omega}{2k_B T}}$):

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(\frac{e^{\frac{\hbar \omega}{k_B T}} - 1}{e^{\frac{\hbar \omega}{2k_B T}}}\right)^2} = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(e^{\frac{\hbar \omega}{2k_B T}} - e^{-\frac{\hbar \omega}{2k_B T}}\right)^2}$$

Using the identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we can write $e^x - e^{-x} = 2\sinh(x)$. Let $x = \frac{\hbar \omega}{2k_B T}$:

$$C_V = k_B \left(2 \cdot \frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\left(2 \sinh\left(\frac{\hbar \omega}{2k_B T}\right)\right)^2}$$

$$C_V = k_B \cdot 4 \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{4 \sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$

Canceling out the factor of 4 yields:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$

---

# 5. Asymptotic Behavior (Discussion)

In the high-temperature limit where $k_B T \gg \hbar \omega$, the parameter $x = \frac{\hbar \omega}{2k_B T}$ becomes very small ($x \to 0$).

We can perform a Taylor expansion for the hyperbolic sine function around zero:

$$\sinh(x) = x + \frac{x^3}{6} + \dots \approx x$$

Substituting this linear approximation back into our expression for $C_V$:

$$C_V \approx k_B \cdot x^2 \cdot \frac{1}{x^2} = k_B$$

### Conclusion
As $T \to \infty$, $C_V \to k_B$.

This matches the classical **equipartition theorem** prediction exactly. A classical 1D harmonic oscillator has two quadratic degrees of freedom: kinetic energy ($\frac{P^2}{2m}$) and potential energy ($\frac{1}{2}m\omega^2 X^2$). Since each quadratic degree of freedom contributes $\frac{1}{2}k_B$ to the heat capacity, the total classical prediction is $2 \times \frac{1}{2}k_B = k_B$. At high temperatures, thermal energy completely washes out the discrete quantum spacing ($\Delta E = \hbar \omega$), and the system behaves classically.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

Answer

# Maximum Entropy Derivation for a Continuous Random Variable

We want to find the probability density function $f(x)$ for a continuous random variable $X \in (-\infty, \infty)$ that maximizes the differential entropy:

$$H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx$$

subject to the following three constraints:
1. **Normalization:** $\int_{-\infty}^{\infty} f(x) \, dx = 1$
2. **Mean:** $\int_{-\infty}^{\infty} x f(x) \, dx = \mu$
3. **Variance:** $\int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx = \sigma^2$

---

### Method 1: Calculus of Variations (Lagrange Multipliers)

We define the functional functional $\mathcal{J}[f]$ using Lagrange multipliers $\lambda_0$, $\lambda_1$, and $\lambda_2$:

$$\mathcal{J}[f] = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} f(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x f(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx - \sigma^2 \right)$$

Combining the integrands:

$$\mathcal{J}[f] = \int_{-\infty}^{\infty} \left[ -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x - \mu)^2 f(x) \right] dx + \lambda_0 + \lambda_1 \mu + \lambda_2 \sigma^2$$

To maximize $\mathcal{J}[f]$, we take the functional derivative with respect to $f(x)$ and set it to zero:

$$\frac{\partial}{\partial f} \left( -f(x) \ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x - \mu)^2 f(x) \right) = 0$$

$$- \ln f(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

Solving explicitly for $\ln f(x)$:

$$\ln f(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2$$

Exponentiating both sides yields the general form:

$$f(x) = \exp\left( -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$

---

### Solving for the Multipliers

#### Step 1: Finding $\lambda_1$
A symmetric shift centered at the mean simplifies the algebra. Notice that the variance constraint is already centered at $\mu$. For the distribution to be symmetric around its mean $\mu$, the linear term in the exponent must vanish when expanded around $\mu$. Thus, we set:
$$\lambda_1 = 0$$

This simplifies our density function to:
$$f(x) = C \exp\left( -\lambda_2 (x - \mu)^2 \right) \quad \text{where } C = e^{-(1 + \lambda_0)}$$

#### Step 2: Finding $C$ (Normalization)
Using the total probability constraint:
$$\int_{-\infty}^{\infty} C \exp\left( -\lambda_2 (x - \mu)^2 \right) \, dx = 1$$

Using the standard Gaussian integral $\int_{-\infty}^{\infty} e^{-a(x-b)^2} \, dx = \sqrt{\frac{\pi}{a}}$:
$$C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$

#### Step 3: Finding $\lambda_2$ (Variance)
Substitute $f(x)$ into the variance constraint:
$$\int_{-\infty}^{\infty} (x - \mu)^2 \sqrt{\frac{\lambda_2}{\pi}} \exp\left( -\lambda_2 (x - \mu)^2 \right) \, dx = \sigma^2$$

Using the known identity $\int_{-\infty}^{\infty} y^2 e^{-\alpha y^2} \, dy = \frac{1}{2\alpha}\sqrt{\frac{\pi}{\alpha}}$:
$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2$$

$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

---

### Final Substitution

Now we substitute $\lambda_2 = \frac{1}{2\sigma^2}$ back into our expression for $C$:
$$C = \sqrt{\frac{1}{2\pi\sigma^2}}$$

Placing $C$ and $\lambda_2$ back into the simplified density function $f(x)$:

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

This is precisely the probability density function of a normally distributed random variable, proving that $X \sim \mathcal{N}(\mu, \sigma^2)$ maximizes differential entropy.

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




Answer

## Problem Formulation & Optimization Solution

We need to solve for the conditional routing probabilities $P(Y=1 \mid \mathbf{x})$ for each observable state vector $\mathbf{x}$. Let these unknown parameters be denoted by:

$$
\begin{aligned}
\theta_A &= P(Y=1 \mid \mathbf{x}_A = [1, 0, 0]^T) \\
\theta_B &= P(Y=1 \mid \mathbf{x}_B = [0, 1, 0]^T) \\
\theta_C &= P(Y=1 \mid \mathbf{x}_C = [0, 0, 1]^T) \\
\theta_D &= P(Y=1 \mid \mathbf{x}_D = [0, 1, 1]^T) \\
\theta_E &= P(Y=1 \mid \mathbf{x}_E = [1, 1, 0]^T)
\end{aligned}
$$

All other states have a marginal probability of $0$, so their conditional probabilities do not affect our expectations. Because these are probabilities, they are strictly bounded by $0 \le \theta_i \le 1$.

---

### Step 1: Formulating the Constraint Equations

Using the definition of expectation for joint events, we expand the constraints into linear equations using the law of total probability $\mathbb{E}[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x}} x_i \cdot P(Y=1 \mid \mathbf{x}) \cdot P(\mathbf{x})$.

#### Constraint 1: Financial Words Routed to Technical
$$\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = (1 \cdot \theta_A \cdot 0.40) + (0 \cdot \theta_B \cdot 0.25) + (0 \cdot \theta_C \cdot 0.05) + (0 \cdot \theta_D \cdot 0.15) + (1 \cdot \theta_E \cdot 0.05) = 0.05$$
$$\implies 0.40\theta_A + 0.05\theta_E = 0.05$$

#### Constraint 2: Account Access Words Routed to Technical
$$\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = (0 \cdot \theta_A \cdot 0.40) + (1 \cdot \theta_B \cdot 0.25) + (0 \cdot \theta_C \cdot 0.05) + (1 \cdot \theta_D \cdot 0.15) + (1 \cdot \theta_E \cdot 0.05) = 0.15$$
$$\implies 0.25\theta_B + 0.15\theta_D + 0.05\theta_E = 0.15$$

#### Constraint 3: Bug/Malfunction Words Routed to Technical
$$\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = (0 \cdot \theta_A \cdot 0.40) + (0 \cdot \theta_B \cdot 0.25) + (1 \cdot \theta_C \cdot 0.05) + (1 \cdot \theta_D \cdot 0.15) + (0 \cdot \theta_E \cdot 0.05) = 0.16$$
$$\implies 0.05\theta_C + 0.15\theta_D = 0.16$$

---

### Step 2: Solving the System of Equations

From **Constraint 3**, we isolate the bounds for $\theta_D$:
$$0.05\theta_C + 0.15\theta_D = 0.16 \implies 0.15\theta_D = 0.16 - 0.05\theta_C$$
Since $\theta_C \le 1$, the minimum value for $0.15\theta_D$ occurs when $\theta_C = 1$:
$$0.15\theta_D \ge 0.16 - 0.05(1) = 0.11 \implies \theta_D \ge \frac{0.11}{0.15} \approx 0.7333$$

Since $\theta_D \le 1$, the maximum value for $0.05\theta_C$ occurs when $\theta_D$ is at its maximum allowable value to satisfy the equation. If we set $\theta_D = 1$:
$$0.05\theta_C + 0.15(1) = 0.16 \implies 0.05\theta_C = 0.01 \implies \theta_C = 0.20$$

Thus, one perfectly valid exact solution vector that satisfies all constraints simultaneously is:

| Parameter | Calculation | Exact Value | Destination $Y=1$ (Technical) | Destination $Y=0$ (Finance) |
| :--- | :--- | :--- | :--- | :--- |
| **$\theta_A$ (Pure Financial)** | $0.05 / 0.40$ | **$0.125$** | $12.5\%$ | $87.5\%$ |
| **$\theta_B$ (Pure Access)** | $0.00 / 0.25$ | **$0.000$** | $0.0\%$ | $100.0\%$ |
| **$\theta_C$ (Pure Technical)** | Given $\theta_D=1$ | **$0.200$** | $20.0\%$ | $80.0\%$ |
| **$\theta_D$ (Access Tech)** | Maximum bound | **$1.000$** | $100.0\%$ | $0.0\%$ |
| **$\theta_E$ (Billing Access)** | Setting to $0$ | **$0.000$** | $0.0\%$ | $100.0\%$ |

---

### Step 3: Verification of the Solution

Let's plug these values back into our expectation equations to verify their analytical correctness:

1. **Financial Constraint:** $(0.40 \times 0.125) + (0.05 \times 0) = 0.05 + 0 = 0.05$ (Valid)
2. **Access Constraint:** $(0.25 \times 0) + (0.15 \times 1.0) + (0.05 \times 0) = 0.15$ (Valid)

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



Answer

## Maximum Entropy (MaxEnt) Asset Price Distribution Formulation & Solution

We want to find the least-biased continuous probability density function $p(S_T)$ for $S_T > 0$ that satisfies our observed market constraints while maximizing Shannon information entropy.

### Step 1: The Optimization Objective

The objective function to maximize is the continuous differential entropy:

$$
\max_{p(S_T)} \mathcal{H}(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T
$$

Subject to the following four structural and market-clearing constraints:

1. **Normalization (Total Probability):**  
   $$\int_{0}^{\infty} p(S_T) \, dS_T = 1$$

2. **Forward Price Constraint ($\mathbb{E}[S_T]$):**  
   $$\int_{0}^{\infty} S_T \cdot p(S_T) \, dS_T = 102$$

3. **Volatility/Variance Constraint ($\mathbb{E}[S_T^2]$):**  
   $$\int_{0}^{\infty} S_T^2 \cdot p(S_T) \, dS_T = 10,600$$

4. **Downside Skew Constraint ($\mathbb{E}[\ln(S_T)]$):**  
   $$\int_{0}^{\infty} \ln(S_T) \cdot p(S_T) \, dS_T = 4.60$$

---

### Step 2: The Method of Lagrange Multipliers

To solve this constrained optimization problem, we construct the Lagrangian functional $\mathcal{L}$ using multipliers $\lambda_0, \lambda_1, \lambda_2, \lambda_3$:

$$
\mathcal{L}(p, \lambda) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) dS_T - \lambda_0 \left( \int_{0}^{\infty} p(S_T) dS_T - 1 \right) - \lambda_1 \left( \int_{0}^{\infty} S_T p(S_T) dS_T - 102 \right) - \lambda_2 \left( \int_{0}^{\infty} S_T^2 p(S_T) dS_T - 10,600 \right) - \lambda_3 \left( \int_{0}^{\infty} \ln(S_T) p(S_T) dS_T - 4.60 \right)
$$

Taking the functional derivative with respect to $p(S_T)$ and setting it to zero via the calculus of variations:

$$
\frac{\delta \mathcal{L}}{\delta p(S_T)} = -\ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0
$$

Solving for $\ln p(S_T)$:

$$
\ln p(S_T) = -(1 + \lambda_0) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)
$$

Taking the exponential of both sides yields the functional form of the optimal probability density function:

$$
p^*(S_T) = \exp(-(1 + \lambda_0)) \cdot \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right)
$$
